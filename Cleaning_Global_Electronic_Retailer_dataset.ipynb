Cleaning Tabel Stores

```python
import numpy as np
import pandas as pd
```

Cleaning Tabel Customers

```python
df_customers = pd.read_csv('/content/Customers.csv', encoding='latin1')
```

```python
df_customers.head()
```

```python
df_customers.dtypes
```

```python
# mengubah nama kolom
df_customers.columns = df_customers.columns.str.replace(' ','_').str.lower()

df_customers.rename(columns={'customerkey':'customer_key'}, inplace=True)

# cek kolom
df_customers.columns
```

```python
# mengubah tipe data tiap-tiap kolom
col_category = ['gender', 'city', 'state_code', 'state', 'zip_code', 'country', 'continent']

for col in col_category:
  df_customers[col] = df_customers[col].astype('category')

df_customers['birthday'] = pd.to_datetime(df_customers['birthday'])
```

```python
# periksa sampel random
df_customers.sample(5)
```

```python
# cek data duplikat
df_customers.duplicated().sum()
```

```python
df_customers.isnull().sum()
```

```python
mask = df_customers['state_code'].isna()
df_customers.loc[mask]
```

```python
# mengisi state code untuk NAPOLY yaitu NA
# menambahkan 'NA' sebagai kategori terlebih dahulu
df_customers['state_code'] = df_customers['state_code'].cat.add_categories('NA')
df_customers.loc[mask, 'state_code'] = 'NA'
```

```python
df_customers.nunique()
```

```python
df_customers.to_csv('customers_cleaned.csv')
```

Cleaning Tabel Products

```python
df_products = pd.read_csv('/content/Products.csv')
```

```python
df_products.info()
```

```python
# mengubah nama kolom
df_products.columns = df_products.columns.str.replace(' ','_').str.lower()
df_products.rename(columns={'productkey':'product_key',
                            'subcategorykey':'subcategory_key',
                            'categorykey':'category_key'},
                   inplace=True)
```

```python
df_products.head()
```

```python
# mengubah tipe data category
cols_cat = ['brand', 'color', 'subcategory', 'category']

df_products[cols_cat] = df_products[cols_cat].astype('category')
```

```python
# mengubah format unit cost dan unit price
df_products['unit_cost_usd'] = (df_products['unit_cost_usd']
                                .astype(str)
                                .str.replace(r'[^0-9.]', '', regex=True)
                                .astype(float))
df_products['unit_price_usd'] = (df_products['unit_price_usd']
                                .astype(str)
                                .str.replace(r'[^0-9.]', '', regex=True)
                                .astype(float))
```

```python
# memeriksa apakah semua nilai valid
df_products.loc[
    (df_products['unit_cost_usd'] <= 0) |
    (df_products['unit_price_usd'] <= 0)
]
```

```python
# membuat kolom baru
df_products['margin'] = df_products['unit_price_usd'] - df_products['unit_cost_usd']
```

```python
# memeriksa apakah margin selalu positif
df_products.loc[
    (df_products['margin'] <= 0) |
    (df_products['margin'] <= 0)
]
```

```python
df_products.isnull().sum()
```

```python
df_products.duplicated().sum()
```

```python
df_products.to_csv('products_cleaned.csv')
```
